## 4. Spatial Join — Assign drinking fountains to districts

### 4.1 Convert coordinates to Points

GeoPandas expects longitude (X), then latitude (Y), with CRS `EPSG:4326`.

```python
# Convert the CSV longitude/latitude columns into Point geometry.
fountains = gpd.GeoDataFrame(
    fountain_table.copy(),
    geometry=gpd.points_from_xy(fountain_table["longitude"], fountain_table["latitude"]),
    crs="EPSG:4326"
)
```

### 4.2 Inspect the new GeoDataFrame

```python
print("Rows:", len(fountains))
print("Geometry types:", fountains.geom_type.unique().tolist())
print("CRS:", fountains.crs)
display(fountains[["fountain_id", "latitude", "longitude", "geometry"]].head())
```

### 4.3 Compare CRS

Both layers must use the same CRS before a spatial join.

```python
print("Fountain CRS:", fountains.crs)
print("District CRS:", districts.crs)
if fountains.crs != districts.crs:
    raise ValueError("CRS mismatch: transform one layer with .to_crs(...) before joining.")
```

### 4.4 Map the inputs — Optional

Show the fountain points on top of district boundaries. The points do not have district IDs yet.


```python
print("Fountain points:", len(fountains))
print("District polygons:", len(districts))
```

```python
# Optional: draw district polygons first, so the fountain points appear on top.
fig, ax = plt.subplots(figsize=(9, 6))
districts.plot(ax=ax, facecolor="#e4eef6", edgecolor="#356489", linewidth=1.4)
fountains.plot(ax=ax, color="#bd643c", markersize=19, alpha=0.85)
# Labels help connect the map to district_id in the tables.
for _, row in districts.iterrows():
    p = row.geometry.representative_point()
    ax.annotate(str(row.district_id), (p.x, p.y), fontsize=11, weight="bold", ha="center")
ax.set_title("Drinking-fountain points and district boundaries")
ax.set_axis_off()
plt.tight_layout()
plt.show()
```

### 4.5 Spatial Join with `sjoin()`

`within` identifies the containing district; `how="left"` retains unmatched fountains.

```python
# Match each point to a containing district; keep points with no match.
assigned = gpd.sjoin(
    fountains,
    districts[["district_id", "district_name", "geometry"]],
    how="left",
    predicate="within"
)
```

### 4.6 Inspect the join result

The spatial join adds district attributes to each matched fountain.

```python
print("Before join:", len(fountains), "points")
print("After join:", len(assigned), "rows")
print("New district fields:", [c for c in ["district_id", "district_name"] if c in assigned.columns])
display(assigned[["fountain_id", "district_id", "district_name"]].head(10))
```

### 4.7 Inspect unmatched points

Unmatched points have a missing `district_id`, not a zero.

```python
unmatched = assigned[assigned["district_id"].isna()].copy()
```

```python
print("Unmatched points:", len(unmatched))
display(unmatched[["fountain_id", "longitude", "latitude"]])
```

### 4.7a Map matched and unmatched points — Optional

Show the result of `sjoin()` on a map. The unmatched points did not fall **within** any of the four polygons in this simplified dataset.

```python
# Optional: use the district_id added by sjoin to separate the two groups.
fig, ax = plt.subplots(figsize=(9, 6))
districts.plot(ax=ax, facecolor="#edf2f6", edgecolor="#4b6c83", linewidth=1.2)
assigned.loc[assigned["district_id"].notna()].plot(
    ax=ax, color="#287a62", markersize=23, label="Matched"
)
if len(unmatched) > 0:
    unmatched.plot(ax=ax, color="#c84e3a", marker="x", markersize=45, label="Unmatched")
ax.legend()
ax.set_title("Fountain points: matched and unmatched after sjoin")
ax.set_axis_off()
plt.tight_layout()
plt.show()
```

### 4.8 Check assignments

Compare input and joined row counts. Duplicate IDs may indicate multiple spatial matches.

```python
print("Input fountains:", len(fountains))
print("Joined rows:", len(assigned))
print("Matched:", assigned["district_id"].notna().sum())
print("Unmatched:", assigned["district_id"].isna().sum())
```

```python
print("One joined row per fountain:", len(assigned) == len(fountains))
print("Duplicate fountain IDs:", assigned["fountain_id"].duplicated().sum())
```

## 5. GroupBy — From points to district totals

### 5.1 Keep matched points

Unmatched points are reported separately.

```python
matched = assigned.dropna(subset=["district_id"]).copy()
```

```python
print("Matched:", len(matched))
print("Unmatched:", len(unmatched))
print("Total input points:", len(fountains))
```

### 5.2 GroupBy

`groupby("district_id").size()` counts matched fountains by district.

```python
# Count matched point rows in each district.
fountain_counts = (
    matched.groupby("district_id")
    .size()
    .reset_index(name="fountain_count")
)
```

```python
print("Rows in count table:", len(fountain_counts))
display(fountain_counts)
```

### 5.3 Merge counts back to districts

Use a left attribute join to keep every district.

```python
# Add the point counts back to all four district polygons.
district_profile = district_population.merge(
    fountain_counts,
    on="district_id",
    how="left"
)
```

```python
print("Rows in final profile:", len(district_profile))
display(district_profile[["district_id", "district_name", "population_2025", "fountain_count"]])
```

### 5.4 Counts of zero and missing values

`fillna(0)` sets an absent count to zero **within this selected sample**. It does not establish that a district has no fountains in reality.

```python
# A missing group count means zero points within this selected sample.
district_profile["fountain_count"] = district_profile["fountain_count"].fillna(0).astype(int)
```

```python
print("District rows:", len(district_profile))
print("Total counted fountains:", district_profile["fountain_count"].sum())
display(district_profile[["district_name", "population_2025", "fountain_count"]])
```

### 5.5 Visualise fountain counts — Optional

The map shows the district totals geographically. The bar chart makes the counts easier to compare. The final plot explores population versus fountain counts; it is **not** a measure of access.


```python
print("Rows:", len(district_profile))
print("CRS:", district_profile.crs)
print("Geometry types:", district_profile.geom_type.unique().tolist())
```

```python
# Optional: plot the sample count in each polygon as a shaded map.
fig, ax = plt.subplots(figsize=(9, 6))
district_profile.plot(
    column="fountain_count", cmap="YlOrBr", legend=True,
    edgecolor="#34526b", linewidth=1.5, ax=ax,
    legend_kwds={"label": "Matched drinking-fountain points"}
)
# Add the count inside each district to read the exact value.
for _, row in district_profile.iterrows():
    p = row.geometry.representative_point()
    ax.text(p.x, p.y, f"{row.district_id}: {row.fountain_count}",
            ha="center", va="center", fontsize=10, weight="bold")
ax.set_title("Drinking-fountain points in the teaching sample")
ax.set_axis_off()
plt.tight_layout()
plt.show()
```

```python
# Optional: the same counts as bars; bar height is easier to compare than map colour.
fig, ax = plt.subplots(figsize=(8.5, 4.6))
chart_data = district_profile.sort_values("district_id")
ax.bar(chart_data["district_id"].astype(str), chart_data["fountain_count"], color="#3d739d")
ax.set_xlabel("District ID")
ax.set_ylabel("Matched drinking-fountain points")
ax.set_title("Fountain counts by district")
plt.tight_layout()
plt.show()
```

**Optional: Compare population with the fountain count.** Each dot is one district. With only four districts and selected facilities, do not use this plot to conclude anything about access or service quality.

```python
# Optional: inspect two columns together rather than mapping them separately.
fig, ax = plt.subplots(figsize=(7.5, 4.8))
ax.scatter(district_profile["population_2025"] / 1000,
           district_profile["fountain_count"], s=65, color="#3d739d")
# Label each observation using its district ID.
for _, row in district_profile.iterrows():
    ax.annotate(f"D{row.district_id}",
                (row.population_2025 / 1000, row.fountain_count),
                xytext=(5, 5), textcoords="offset points")
ax.set_xlabel("Population (thousands, 2025)")
ax.set_ylabel("Matched fountain points")
ax.set_title("Population and fountain counts in four districts")
plt.tight_layout()
plt.show()
```
